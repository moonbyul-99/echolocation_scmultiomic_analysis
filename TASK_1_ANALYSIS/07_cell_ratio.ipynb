```python
import numpy as np 
import pandas as pd 
import scanpy as sc 
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
data_path = '../data/final_anno.h5ad'

scdata = sc.read_h5ad(data_path)
scdata
```

```python
save_dir = 'cell_ratio'
os.makedirs(save_dir, exist_ok=True)
```

## PLOT THE CELL TYPE RATIO DIFFERENCE

```python
adata = scdata[~scdata.obs.species.isin(['M', 'QF'])]
df = pd.crosstab(adata.obs.tissues, adata.obs.celltype)
df.to_csv(os.path.join(save_dir, 'eco_cell_num.csv'))
df_sum = df.sum(axis = 1)
e_df = df.T/df_sum
e_df.to_csv(os.path.join(save_dir, 'eco_cell_ratio.csv'))
```

```python
adata = scdata[scdata.obs.species.isin(['M', 'QF'])]
df = pd.crosstab(adata.obs.tissues, adata.obs.celltype)
df.to_csv(os.path.join(save_dir, 'neco_cell_num.csv'))
df_sum = df.sum(axis = 1)
n_df = df.T/df_sum
n_df.to_csv(os.path.join(save_dir, 'neco_cell_ratio.csv'))
```

```python
assert (e_df.index == n_df.index).all()
assert (e_df.columns == n_df.columns).all()
```

```python
sns.heatmap(e_df - n_df ,cmap = 'RdBu_r')
plt.savefig(os.path.join(save_dir, 'heatmap_diff.pdf'), bbox_inches='tight')
plt.show()
```
