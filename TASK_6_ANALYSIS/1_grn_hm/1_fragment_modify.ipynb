## jthm modify

```python
df = pd.read_csv('/home/rsun@ZHANGroup.local/sly_data/grn_momics/data/filter_fragments/jthm_fragments/fragment.tsv', sep='\t', header = None)
df.head()
```

```python
cell_id = df.iloc[:,3].values
tmp = []
for ele in cell_id:
    a,b,c = ele.split('-')
    tmp.append('-'.join([a,b]))
df.iloc[:,3] = tmp
df.head()
```
