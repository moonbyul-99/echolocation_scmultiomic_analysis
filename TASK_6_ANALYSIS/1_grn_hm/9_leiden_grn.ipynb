```python
import os
import subprocess

for i in range(1, 11):  # 示例：循环创建 leiden_0_pipeline 到 leiden_10_pipeline
    dir_name = f'leiden_grn/leiden_{i}_pipeline'
    os.makedirs(dir_name, exist_ok=True)

    result = subprocess.run(
        ["scenicplus", "init_snakemake", "--out_dir", dir_name],
        capture_output=True,
        text=True,
        check=True
    )
    os.makedirs(os.path.join(dir_name, "outs"))
    os.makedirs(os.path.join(dir_name, "tmp"))
```
